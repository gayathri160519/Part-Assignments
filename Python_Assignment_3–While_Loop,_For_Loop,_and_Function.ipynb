```python
# Task 1: While Loop & Control Statements -Number Guessing Game

# 1. Set Up the Game:
# Generate a random number between 1 and 10 that the user has to guess.
# Import random and use randint function.

import random
secret_number = random.randint(1, 10)                                            # random.randint(1, 10) → generates the secret number.
print("Welcome to the Number Guessing Game!")
print("Guess the secret number between 1 and 10.")
while True:                                                                      # while True → keeps the game running until the player guesses correctly.
    try:
        guess = int(input("Enter your guess: "))


        if guess < 1 or guess > 10:
            print(" Please enter a number between 1 and 10.")
            continue                                                             # continue → skips invalid inputs or out-of-range guesses.


        if guess == secret_number:
            print(" Congratulations! You guessed it right!")
            break                                                                # break → exits the loop when the correct number is guessed.


        else:                                                                    # else → runs if the loop ends naturally (without break).
            print(" Wrong guess, try again!")

    except ValueError:
        print(" Invalid input! Please enter a number only.")
        continue
else:

    print("Game Over. Thanks for playing!")
```

```python
# 2. Prompt the User:
# Ask the user to guess the number.
# Set a variable attempts to 3, which represents the maximum number of guesses allowed.

import random
secret_number = random.randint(1, 10)
print(" Welcome to the Number Guessing Game!")
print("You have 3 attempts to guess the secret number between 1 and 10.")
attempts = 3                                                                          # attempts = 3 → limits the number of guesses.
while attempts > 0:
    try:
        guess = int(input("Enter your guess: "))


        if guess < 1 or guess > 10:
            print(" Please enter a number between 1 and 10.")
            continue


        if guess == secret_number:
            print(" Congratulations! You guessed it right!")
            break
        else:
            attempts -= 1                                                           # attempts -= 1 → decreases attempts after each wrong guess.
            print(f" Wrong guess! Attempts left: {attempts}")

    except ValueError:
        print(" Invalid input! Please enter a number only.")
        continue
else:                                                                              # else block → runs when all attempts are used without guessing correctly.

    print(f" Game Over! The secret number was {secret_number}.")
```

```python
# 3. Implement the Guessing Logic:
# Use a while loop to allow the user to keep guessing until they get the correct number or run out of attempts.
# Provide feedback to the user for each guess:
# If the guess is out of the valid range (1 to 10), inform the user.
# If the guess is greater than the secret number.
# If the guess is lower than the secret number.
# If the guess is correct, congratulate the user and end the game.

import random
secret_number = random.randint(1, 10)                                                   # Range check → warns if guess is outside 1–10.
print(" Welcome to the Number Guessing Game!")
print("You have 3 attempts to guess the secret number between 1 and 10.")
attempts = 3
while attempts > 0:
    try:
        guess = int(input("Enter your guess: "))


        if guess < 1 or guess > 10:
            print(" Please enter a number between 1 and 10.")
            continue


        if guess == secret_number:
            print(" Congratulations! You guessed it right!")
            break                                                                 # Equal to secret number → congratulates and ends the game with break.
        elif guess > secret_number:
            attempts -= 1
            print(f" Too high! Attempts left: {attempts}")                        # Greater than secret number → feedback: "Too high!".
        else:
            attempts -= 1
            print(f" Too low! Attempts left: {attempts}")                         # Less than secret number → feedback: "Too low!".

    except ValueError:                                                            # Else block → triggers if all attempts are used without success.
        print(" Invalid input! Please enter a number only.")
        continue
```

```python
# 4. Control Statements:
# Use continue to skip the rest of the loop after informing the user if the guess is out of range.
# Use break to exit the loop when the guess is correct.
# Use else with the while loop to provide a message like "Better luck next time!"
# if the user runs out of attempts without guessing the correct number.

import random
secret_number = random.randint(1, 10)
print(" Welcome to the Number Guessing Game!")
print("You have 3 attempts to guess the secret number between 1 and 10.")
attempts = 3
while attempts > 0:
    try:
        guess = int(input("Enter your guess: "))


        if guess < 1 or guess > 10:
            print(" Please enter a number between 1 and 10.")
            continue                                                               # continue → skips the rest of the loop if the guess is out of range or invalid.


        if guess == secret_number:
            print(" Congratulations! You guessed it right!")
            break                                                                 # break → exits the loop immediately when the guess is correct.


        elif guess > secret_number:
            attempts -= 1
            print(f" Too high! Attempts left: {attempts}")
        else:                                                                    # else (with while loop) → executes only if the loop finishes without break (i.e., all attempts are used).
            attempts -= 1
            print(f" Too low! Attempts left: {attempts}")

    except ValueError:
        print(" Invalid input! Please enter a number only.")
        continue
else:

    print(" Better luck next time!")
    print(f"The secret number was {secret_number}.")
```

```python
# Task 2:- For Loop - Multiplication Table Generator

# 1. Prompt user for input. Ask the user to enter a number for which they want to generate a multiplication table.


number = int(input("Enter a number to generate its multiplication table: "))       # input() → asks the user to enter a number.
print(f"\n Multiplication Table for {number}\n")                                   # int() → converts the input into an integer.
for i in range(1, 11):                                                            # for i in range(1, 11) → loops from 1 to 10.
    result = number * i
    print(f"{number} x {i} = {result}")                                            # print() → displays each multiplication result in a neat format.
```

```python
# Task 3: Function- BMI Calculator

# Problem Statement: Create a Python program that calculates the Body Mass Index (BMI).
# Hint: BMI = weight (kg) / [height (m)]²
# Instructions:
# 1. Define a function calculate_bmi(weight, height) that returns the BMI.
# 2.  Prompt the user for their weight (in kg) and height (in meters).
# 3.  Use the function to calculate and display the BMI


def calculate_bmi(weight, height):                                                  # Function calculate_bmi(weight, height) → computes BMI using the formula.
    return weight / (height ** 2)                                                   # Calculation → divides weight by height squared.
weight = float(input("Enter your weight in kilograms: "))                           # User input → asks for weight (kg) and height (m).
height = float(input("Enter your height in meters: "))
bmi = calculate_bmi(weight, height)
print(f"\n Your BMI is: {bmi:.2f}")                                                 # Output → prints BMI rounded to 2 decimal places.

```
