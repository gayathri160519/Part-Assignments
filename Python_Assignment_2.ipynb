```python
# Python Assignment 2: Data Structures - List, Dictionary, Set & Conditional Statements

# 1. List Creation:

# a. Create a list named age_list with five integer elements
age_list = [24, 25, 26, 27, 28]

# b. Create a list named name_list with five string elements
name_list = ["Alice", "Bob", "Charlie", "David", "Eva"]

print(age_list)                                                     #It prints the age_list and name_list
print(name_list)
```

```python
# 2.  List Operations / Modifications:

# Initial lists
age_list = [24, 25, 26, 27, 28]
name_list = ["Alice", "Bob", "Charlie", "David", "Eva"]

# a. Append "Yazhini" to name_list                                 # adds Yazhini at the end of the list.
name_list.append("Yazhini")

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)                                            # inserts 30 at the index number 2 in age_list.

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")                                       # removes Yazhini in the name_list.

# d. Pop the last element from age_list
age_list.pop()                                                     # removes 28 in the age_list.

# e. Extend age_list with additional ages [29, 30, 26]
age_list.extend([29, 30, 26])                                     # adds 29,30,26 in the age_list.

# f. Sort age_list in descending order
age_list.sort(reverse=True)                                       # age_list is in descending order.

# g. Find Max age, Min age, and sum of all ages
max_age = max(age_list)                                            # gives maximum
min_age = min(age_list)                                            # gives minimum
sum_age = sum(age_list)                                            # gives sum

print("Final age_list:", age_list)
print("Final name_list:", name_list)
print("Max age:", max_age)
print("Min age:", min_age)
print("Sum of ages:", sum_age)
```

```python
# 3. Accessing List Elements:

# Example list
name_list = ["Alice", "Bob", "Charlie", "David", "Eva"]

# a. Print the first element of name_list
print("First element:", name_list[0])                         # Alice

# b. Print the last element of name_list
print("Last element:", name_list[-1])                         # Eva -> -1 refers Last Element.

# c. Print the elements from index 2 to index 4
print("Index 2 to 4:", name_list[2:5])                        # ['Charlie', 'David', 'Eva'] -> Slicing

# d. Print the elements of name_list in reverse order
print("Reverse order:", name_list[::-1])                      # ['Eva', 'David', 'Charlie', 'Bob', 'Alice'] -> [::-1]-> reverses the List.
```

```python
# Sets (Operations):
# a. Create a set called my_set with following values: ['a','e','i','o','u','a','a','i']

my_set = set(['a','e','i','o','u','a','a','i'])           # A set in Python is an unordered collection of unique elements.
print(my_set)                                             # All duplicate values ('a' appears multiple times, 'i' appears twice) are removed automatically.
                                                          # Only unique elements remain.
```

```python
# b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.

my_set = {'a','e','i','o','u'}                       # Sets are unordered collections of unique elements.
my_set[4] = 's'                                      # Unlike lists, sets do not support indexing (my_set[4] is invalid).
                                                     # Because there’s no fixed position for elements, we cannot assign a new value at a specific index.
```

```python
my_set.add('s')                                  # Sets only allow operations like:
print(my_set)                                    # add() → to insert a new element.
                                                 # remove() / discard() → to delete an element.
                                                 # update() → to add multiple elements.
```

```python
# c. Create two sets: set1 with values: {1, 3, 5, 7, 9}  set2 with values: {2, 3, 5, 8, 10}

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set1:", set1)                           # Both sets contain unique values only.
print("Set2:", set2)                           # Order is not guaranteed when printing (Python may display them in a different sequence).
```

```python
# d. Compute and print the union and intersection of set1 and set2.

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Union → all unique elements from both sets
union_set = set1 | set2                                       # Union (|) → combines both sets, keeping only unique values.
print("Union:", union_set)

# Intersection → common elements between both sets
intersection_set = set1 & set2                                # Intersection (&) → gives the elements that are present in both sets.
print("Intersection:", intersection_set)
```

```python
# Operators & Conditional Statements : (IF, ELIF, ELSE) Performance Category Program:

# 1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

score = int(input("Enter your score (0 to 10): "))                 # input() → takes user input as a string. int() → converts it into an integer.
if 0 <= score <= 10:                                               # The if condition checks whether the score is between 0 and 10 inclusive.
    print("Your score is:", score)                                 # If valid → prints the score.
else:                                                              # If invalid → shows an error message.
    print("Invalid score! Please enter a value between 0 and 10.")
```

```python
# 2. Find the performance category based on the input score using following criteria:
# a. Above Average:  Score greater than 7
# b. Average: Score between 4 and 7(both inclusive)
# c. Below Average: Score lesser than 4

score = int(input("Enter your score (0 to 10): "))
if 0 <= score <= 10:
    if score > 7:
        print("Performance Category: Above Average")                               # Print the Performance category
    elif 4 <= score <= 7:
        print("Performance Category: Average")
    else:
        print("Performance Category: Below Average")
else:
    print("Invalid score! Please enter a value between 0 and 10.")
```
