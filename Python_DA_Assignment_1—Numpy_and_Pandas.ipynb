```python
# Numpy Array Operations:
# Scenario: You are given a list of daily average temperatures (in degrees Celsius) recorded over two weeks. Your task is to use NumPy to analyze this data.
# 1. Create a 1D NumPy Array:
# Create a 1D NumPy array named temperatures_w1 for Week 1 with the following values:  [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

import numpy as np

# Create a 1D NumPy array for Week 1 temperatures

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])           # converts your Python list into a NumPy array.
print("Week 1 Temperatures:", temperatures_w1)
```

```python
# 2. Inspection and Properties:
# Inspect the shape, data type, and the number of elements in the array. Inspect the shape of the temperatures_w1 array.

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Inspect properties

print("Array:", temperatures_w1)
print("Shape:", temperatures_w1.shape)                                           # (7,) → 1D array with 7 elements
print("Data Type:", temperatures_w1.dtype)                                       # float64 → since values are decimals
print("Number of Elements:", temperatures_w1.size)                               # 7
```

```python
# 3. Array Operations:
# ● Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: Fahrenheit = (Celsius * 9/5) + 32
# ● Find the maximum, minimum, and mean temperatures for the week.

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])           # Week 1 temperatures in Celsius

# Convert to Fahrenheit

temperatures_w1_f = (temperatures_w1 * 9/5) + 32

# Find max, min, mean

max_temp = np.max(temperatures_w1)                                               # np.max() → finds the hottest day
min_temp = np.min(temperatures_w1)                                               # np.min() → finds the coldest day
mean_temp = np.mean(temperatures_w1)                                             # np.mean() → computes the average temperature across the week

print("Temperatures in Fahrenheit:", temperatures_w1_f)
print("Maximum Temperature (°C):", max_temp)
print("Minimum Temperature (°C):", min_temp)
print("Mean Temperature (°C):", mean_temp)
```

```python
# 4. Array Slicing and Indexing:
# ● Extract the temperatures for the first three days of the week.
# ● Extract the temperatures for the weekend (last two days).
# ● Extract the temperatures for the middle three days of the week.

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# First three days (indexes 0,1,2)

first_three_days = temperatures_w1[:3]                                           # [:3] → slice from start up to index 3 (exclusive), giving elements at positions 0,1,2.

# Weekend (last two days → indexes 5,6)

weekend = temperatures_w1[-2:]                                                   # [-2:] → slice the last two elements of the array.

# Middle three days (indexes 2,3,4)

middle_three_days = temperatures_w1[2:5]                                         # [2:5] → slice from index 2 up to 5 (exclusive), giving elements at positions 2,3,4.

print("First three days:", first_three_days)
print("Weekend:", weekend)
print("Middle three days:", middle_three_days)
```

```python
# 5. Create a 2D Array:
# Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:
# ★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
# ★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

import numpy as np

# Week 1 and Week 2 temperatures

week1 = [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]
week2 = [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]

# Create 2D array

temperatures = np.array([week1, week2])
print("2D Temperatures Array:\n", temperatures)
print("Shape:", temperatures.shape)                                               # (2, 7) → 2 rows (weeks), 7 columns (days)
                                                                                  # Shape (2, 7) means 2 rows × 7 columns.
```

```python
# 6.  Inspect and Slice the 2D Array:
# ● Inspect and print the shape, data type, and total number of elements in the temperatures array.
# ● Extract the temperatures for each week and the weekends (last two days) for both weeks.

import numpy as np

# 2D array with Week 1 and Week 2 temperatures

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],                                  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]                                   # Week 2
])

# Inspect properties

print("Shape:", temperatures.shape)                                              # (2, 7) → 2 weeks, 7 days each
print("Data Type:", temperatures.dtype)                                          # float64 -> shows the type of values
print("Total Elements:", temperatures.size)                                      # 14

# Extract each week

week1 = temperatures[0]                                                          # Row 0
week2 = temperatures[1]                                                          # Row 1

# Extract weekends (last two days for both weeks)

weekends = temperatures[:, -2:]                                                  # temperatures[:, -2:] → selects all rows (:) and the last two columns (-2:), giving weekend values for both weeks.

print("Week 1 Temperatures:", week1)
print("Week 2 Temperatures:", week2)
print("Weekend Temperatures (both weeks):\n", weekends)
```

```python
# 1. Creating Pandas Series:
# ● Create a Pandas Series named marks with the following data:
# ➢ Marks: 95, 92, 89, 85, 80
# ➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

import pandas as pd

# Creating the Series

marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print(marks)
```

```python
# 2. Indexing and Slicing:
# ●  Use the integer index position to access the mark of the 1st rank student.
# ●  Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
# ● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

import pandas as pd

marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

# 1. Use integer index position to access the mark of the 1st rank student

print("Mark of 1st rank (integer index):", marks[0])                             # Access by integer position.

# 2. Use loc accessor to retrieve the marks of the top 3 ranks

print("\nMarks of top 3 ranks (loc):")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])                                    # Access by labels.

# 3. Use iloc accessor to retrieve the mark of the 3rd-ranked student

print("\nMark of 3rd rank (iloc):", marks.iloc[2])                               # Access by integer position (3rd element).

# 4. Apply a boolean mask to filter ranks where marks > 90

print("\nRanks with marks > 90:")
print(marks[marks > 90])
```

```python
# 3. Manipulating Series:
# ● Modify the mark of the student with the 1st rank to 100.
# ● Remove the entry corresponding to the last rank from the Series.
# ● Compute the CGPA by dividing each mark in the Series by 10.

import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

# 1. Modify the mark of the student with the 1st rank to 100

marks['Rank1'] = 100                                                             # Updates the value.
print("After updating Rank1 to 100:\n", marks)

# 2. Remove the entry corresponding to the last rank

marks = marks.drop('Rank5')                                                      # Removes the last rank entry.
print("\nAfter removing Rank5:\n", marks)

# 3. Compute the CGPA by dividing each mark by 10

cgpa = marks / 10
print("\nCGPA values:\n", cgpa)
```

```python
# 1. Creating Pandas DataFrame:
# ● Create a Pandas DataFrame named transactions with the following data: ➢
# TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
# ➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
# ➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'
# ➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

import pandas as pd

# Define the data

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

# Create the DataFrame

transactions = pd.DataFrame(data)

# Display the DataFrame

print(transactions)
```

```python
# 2. Data Exploration:
# ● Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.
# ● Display only the 'ProductCategory' and 'Amount' columns.
# ● Retrieve the last 3 columns of the DataFrame using iloc or loc.
# ● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
# ● Find the value counts for the 'ProductCategory' column.
# ● Find the unique values in the 'Region' column.
# ● Group by 'Region' and find the mean amount for each region.

import pandas as pd

# DataFrame creation (from step 1)
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],          # Shape → (10, 4) (10 rows, 4 columns)
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

# 1. Display DataFrame

print("Transactions DataFrame:\n", transactions)

# 2. Basic info

print("\nHead:\n", transactions.head())                                          # head() → shows the first 5 rows of the DataFrame by default.TransactionID 101–105.
print("\nTail:\n", transactions.tail())                                          # tail() → shows the last 5 rows of the DataFrame by default.TransactionID 106–110.
print("\nShape:", transactions.shape)
print("\nColumn Names:", transactions.columns.tolist())
print("\nData Types:\n", transactions.dtypes)

# 3. Display only ProductCategory and Amount

print("\nProductCategory & Amount:\n", transactions[['ProductCategory', 'Amount']])

# 4. Retrieve last 3 columns using iloc

print("\nLast 3 Columns (iloc):\n", transactions.iloc[:, -3:])                   # iloc → stands for index location. It lets you select rows and columns by their numerical index (position), not by name.
                                                                                 # : → means "select all rows". -3: → means "select the last 3 columns".


# 5. Filter rows where Region is 'North' and Amount > 200

print("\nFiltered Rows (Region=North & Amount>200):\n",
      transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

# 6. Value counts for ProductCategory

print("\nValue Counts (ProductCategory):\n", transactions['ProductCategory'].value_counts())   # Value Counts (ProductCategory) → Electronics: 4, Clothing: 3, Furniture: 3

# 7. Unique values in Region

print("\nUnique Regions:\n", transactions['Region'].unique())                     # Unique Regions → ['North', 'South', 'East', 'West']

# 8. Group by Region and mean Amount

print("\nMean Amount by Region:\n", transactions.groupby('Region')['Amount'].mean())  # This tells Pandas: “Group the rows of the DataFrame based on the values in the Region column.”
                                                                                      # So all transactions from North are grouped together, all from South together, etc.
                                                                                      # .mean()-> This calculates the average value of Amount for each region group.
```

```python
# 3. Manipulating the DataFrame:
# ●  Modify the 'Amount' for the transaction with TransactionID 102 to 165.
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'.
# ●  Remove the row with TransactionID 109.
# ●  Delete the 'Discount' column from the DataFrame.

import pandas as pd

# Original DataFrame

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

# 1. Modify the 'Amount' for TransactionID 102 to 165
                                                                                 # Used .loc to target the row where TransactionID == 102 and updated Amount.
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165           # It lets you select rows and columns by their labels (names), not by position.

# 2. Add a new column 'Discount' (10% of Amount)

transactions['Discount'] = transactions['Amount'] * 0.10                         # Add Discount column → Simple calculation: Amount * 0.10.

# 3. Remove the row with TransactionID 109

transactions = transactions[transactions['TransactionID'] != 109]                # This checks each row’s TransactionID and returns True if it is not equal to 109, otherwise False.

# 4. Delete the 'Discount' column

transactions = transactions.drop(columns=['Discount'])                           # Delete Discount column → Used .drop(columns=['Discount']).

# Display final DataFrame

print(transactions)
```
