```python
#Strings (Concatenation, Slicing, and Other Methods) :                            # Used AI Support to clear errors and for coding idea #
# 1. String Concatenation:
# input from the user
name = input("Enter your Name: ")

# Concatenate with "Hello"
greeting = "Hello " + name
print(greeting)

# Concatenate with another string
output = greeting + ", welcome to Python programming"
print(output)
```

```python
# 2. String Slicing and Indexing:
# Concatenated string
text = "Hello Gayathri, welcome to Python programming"

# a. First character
print("First character:", text[0])

# b. Last character
print("Last character:", text[-1])

# c. First 5 characters
print("First 5 characters:", text[:5])

# d. Last 11 characters
print("Last 11 characters:", text[-11:])

# e. Reverse the string
print("Reversed string:", text[::-1])

# f. Extract the word "Python"
print("Word 'Python':", text[27:33])
```

```python
#3.  String Methods:
# Input string
strM = "Python beginner tutorial"

# a. Convert to uppercase
print("Uppercase:", strM.upper())

# b. Convert to lowercase
print("Lowercase:", strM.lower())

# c. Capitalize (first letter uppercase, rest lowercase)
print("Capitalize:", strM.capitalize())

# d. Count occurrences of 't'
print("Occurrences of 't':", strM.count('t'))

# e. Replace "Python" with "Data Analytics"
new_str = strM.replace("Python", "Data Analytics")
print("Replaced string:", new_str)
```

```python
#Tuples (Creation, Modification and Access) :
#Create the 1st tuple with values -> (10, 20, 30) and the 2nd tuple with values -> (40, 50, 60):
# Create tuples

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print("Concatenated tuple:", t_combine)

# b. Repeat the elements of t_combine 3 times
t_repeat = t_combine * 3
print("Repeated tuple:", t_repeat)

# c. Access the 3rd element from t_combine
print("3rd element:", t_combine[2])

# d. Access the first three elements from t_combine
print("First 3 elements:", t_combine[:3])

# e. Access the last three elements from t_combine
print("Last 3 elements:", t_combine[-3:])
```
